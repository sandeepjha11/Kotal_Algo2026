```python
import datetime as dt
import pandas as pd
from dateutil.relativedelta import relativedelta

def get_all_expiries(df, index):
    # Filter for symbol and instrument type
    df = df[(df['pSymbolName'] == index) & (df['pInstType'] == 'OPTIDX')]
    
    # Convert epoch to datetime, shift +10 years, subtract 1 day
    df['pExpiryDate'] = df['pExpiryDate'].apply(
        lambda x: dt.datetime.fromtimestamp(x).date() + relativedelta(years=10) - pd.Timedelta(days=1)
    )
    
    # Get unique sorted expiry dates
    all_expiries = sorted(set(df['pExpiryDate']))
    
    # Convert to DataFrame with one expiry per row
    expiry_df = pd.DataFrame(all_expiries, columns=['ExpiryDate'])
    return expiry_df

# Usage
expiry_rows = get_all_expiries(df, 'NIFTY')
print(expiry_rows)
```
